# Mountain NER demo

Run `python train_model.py` first (or point `MODEL` to a Hugging Face Hub id) so that the fine-tuned model exists in `model/mountain_ner_bert/`.

```python
from inference import DEFAULT_MODEL_DIR, load_pipeline, predict_mountains

MODEL = DEFAULT_MODEL_DIR
ner = load_pipeline(MODEL)
```

## Examples

```python
examples = [
    "Mount Rinjani is the second highest volcano in Indonesia.",
    "I dream of climbing K2 one day.",
    "The view from Donguzorun is breathtaking.",
    "We hiked to Cerro Chaltén and then drove to Fitz Roy.",
    "Denali is the highest peak in North America.",
    "Hoverla is the highest point of the Ukrainian Carpathians.",
    "We spent the weekend kayaking down the Danube.",
    "She reached the peak of her career in Paris.",
]

for sentence in examples:
    found = predict_mountains(sentence, ner)
    print(f"{sentence}\n  -> {[m['mountain'] for m in found]}")
```

## Try your own sentence

```python
text = "Last summer we tried to climb Mont Blanc, but the weather near Matterhorn was terrible."
predict_mountains(text, ner)
```
